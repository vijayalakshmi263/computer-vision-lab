```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload video
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Read video
cap = cv2.VideoCapture(filename)

if not cap.isOpened():
    print("Video not found!")

else:
    frames = []

    # Extract all frames
    while True:
        ret, frame = cap.read()

        if not ret:
            break

        frames.append(frame)

    cap.release()

    # Reverse the frames
    frames.reverse()

    # Save reversed video
    height, width = frames[0].shape[:2]

    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    output = cv2.VideoWriter(
        "reverse_video.mp4",
        fourcc, 20, (width, height)
    )

    for frame in frames:
        output.write(frame)

    output.release()

    print("Video reversed successfully!")

    # Display first reversed frame
    plt.figure(figsize=(8, 5))
    plt.imshow(cv2.cvtColor(frames[0], cv2.COLOR_BGR2RGB))
    plt.title("Reversed Video - First Frame")
    plt.axis("off")
    plt.show()

    # Download reversed video
    files.download("reverse_video.mp4")
```
