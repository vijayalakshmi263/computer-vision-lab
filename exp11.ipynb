```python
import numpy as np
from scipy.stats import entropy

# Loan applicant dataset
low_risk = 6
high_risk = 4

# Calculate probabilities
total = low_risk + high_risk
probabilities = [low_risk / total, high_risk / total]

# Calculate entropy
result = entropy(probabilities, base=2)

print("Low Risk Applicants:", low_risk)
print("High Risk Applicants:", high_risk)
print("Entropy:", round(result, 3))

if result == 0:
    print("No uncertainty in classification")
else:
    print("Uncertainty exists in classification")
```
