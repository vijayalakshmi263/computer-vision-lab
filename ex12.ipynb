```python
import cv2
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/content/elephant.jpg')

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.title("Original Elephant Image")
plt.axis("off")
plt.show()
```

```python
rotated_img = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

plt.imshow(rotated_img)
plt.title("270 Degree Clockwise Rotation")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite(
    '/content/rotated_elephant.jpg',
    cv2.cvtColor(rotated_img, cv2.COLOR_RGB2BGR)
)

print("Image saved successfully!")
```
